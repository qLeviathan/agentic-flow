### R Markdown Chunk Options

Common options used:

- `echo=TRUE/FALSE`: Show/hide code
- `eval=TRUE/FALSE`: Execute/skip code
- `warning=TRUE/FALSE`: Show/hide warnings
- `message=TRUE/FALSE`: Show/hide messages
- `include=TRUE/FALSE`: Include/exclude output
- `fig.width`, `fig.height`: Plot dimensions
