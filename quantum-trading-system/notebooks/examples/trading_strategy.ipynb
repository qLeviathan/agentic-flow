## Initialize Strategy

Create a Fibonacci-based trading strategy.

```python
# Initialize Fibonacci strategy
strategy = FibonacciStrategy(
    retracement_levels=[382, 500, 618],  # Integers: 38.2%, 50%, 61.8%
    position_size=1000  # Integer position size
)

print("Strategy Configuration:")
print(f"  Retracement Levels: {strategy.retracement_levels}")
print(f"  Position Size: {strategy.position_size}")
print(f"  Integer-Only: ✅")
```

## Sample Data

Create integer price data for testing.

```python
# Sample price data (all integers)
price_data = [
    100, 105, 108, 103, 102, 107, 112, 110, 115, 118,
    116, 120, 125, 122, 119, 124, 128, 131, 127, 130
]

print(f"Sample Prices (n={len(price_data)}):")
print(price_data)
print(f"\nMin: {min(price_data)}")
print(f"Max: {max(price_data)}")
print(f"Range: {max(price_data) - min(price_data)}")
print(f"\n✅ All values are integers: {all(isinstance(p, int) for p in price_data)}")
```

## Generate Signals

Generate buy/sell signals based on Fibonacci levels.

```python
# Generate trading signals
signals = strategy.generate_signals(price_data)

print("Trading Signals:")
print("=" * 60)
for i, signal in enumerate(signals[:10]):  # Show first 10
    if signal != 0:
        signal_type = "BUY" if signal > 0 else "SELL"
        print(f"Index {i}: {signal_type} - Price: {price_data[i]}")

buy_signals = sum(1 for s in signals if s > 0)
sell_signals = sum(1 for s in signals if s < 0)
print(f"\nTotal Buy Signals: {buy_signals}")
print(f"Total Sell Signals: {sell_signals}")
```

## Risk Management

Apply integer-based risk management.

```python
# Initialize risk manager
risk_mgr = RiskManager(
    max_position_size=10000,  # Integer
    max_loss_per_trade=500,    # Integer
    max_portfolio_risk=2000    # Integer
)

# Calculate position size
entry_price = 100
stop_loss = 95
position_size = risk_mgr.calculate_position_size(
    entry_price,
    stop_loss
)

print("Risk Management:")
print(f"  Entry Price: {entry_price}")
print(f"  Stop Loss: {stop_loss}")
print(f"  Risk per Unit: {entry_price - stop_loss}")
print(f"  Position Size: {position_size}")
print(f"  Total Risk: {position_size * (entry_price - stop_loss)}")
print(f"\n✅ All calculations are integers")
```

## Performance Metrics

Calculate integer-based performance metrics.

```python
# Sample trades
trades = [
    {'entry': 100, 'exit': 110, 'size': 10},  # +100 profit
    {'entry': 105, 'exit': 103, 'size': 10},  # -20 loss
    {'entry': 108, 'exit': 115, 'size': 10},  # +70 profit
    {'entry': 112, 'exit': 108, 'size': 10},  # -40 loss
    {'entry': 110, 'exit': 120, 'size': 10},  # +100 profit
]

# Calculate metrics
total_profit = 0
winning_trades = 0
losing_trades = 0

for trade in trades:
    pnl = (trade['exit'] - trade['entry']) * trade['size']
    total_profit += pnl
    if pnl > 0:
        winning_trades += 1
    elif pnl < 0:
        losing_trades += 1

total_trades = len(trades)
win_rate = (winning_trades * 100) // total_trades  # Integer division

print("Performance Summary:")
print("=" * 60)
print(f"Total Trades: {total_trades}")
print(f"Winning Trades: {winning_trades}")
print(f"Losing Trades: {losing_trades}")
print(f"Win Rate: {win_rate}%")
print(f"Total Profit: {total_profit}")
print(f"\n✅ All metrics are integers")
```
