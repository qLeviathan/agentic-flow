# Basic Data Analysis Example

This notebook demonstrates basic data analysis using the Quantum Trading System's integer-only framework.

## Setup and Imports

Import required modules for analysis.

```python
import sys
import os

# Add project root to path
sys.path.insert(0, os.path.join(os.path.dirname(os.getcwd()), '..'))

from src.encoders.fibonacci_encoder import FibonacciEncoder
from src.encoders.lucas_encoder import LucasEncoder
from src.data.validators.data_validator import DataValidator

print("✅ Imports successful")
print("Framework: Integer-Only Quantum Trading System")
```

## Initialize Encoders

Create Fibonacci and Lucas encoders for integer-based operations.

```python
# Initialize encoders
fib_encoder = FibonacciEncoder(max_n=50)
lucas_encoder = LucasEncoder(max_n=50)

# Display first 10 Fibonacci numbers
fib_sequence = [fib_encoder.get_fibonacci(i) for i in range(10)]
print("Fibonacci (first 10):", fib_sequence)

# Display first 10 Lucas numbers
lucas_sequence = [lucas_encoder.get_lucas(i) for i in range(10)]
print("Lucas (first 10):", lucas_sequence)

# Verify all are integers
all_integers = all(isinstance(n, int) for n in fib_sequence + lucas_sequence)
print(f"\n✅ All values are integers: {all_integers}")
```

## Data Validation

Validate that all operations maintain integer precision.

```python
# Create sample integer data
sample_prices = [100, 105, 103, 108, 112, 110, 115]

# Validate integers
validator = DataValidator()
validation_result = validator.validate_integer_array(sample_prices)

print("Sample Prices:", sample_prices)
print("Validation Result:", validation_result)
print(f"✅ All prices are integers: {validation_result['all_integers']}")
```

## Time Encoding Example

Encode time intervals using Lucas numbers.

```python
from datetime import datetime

# Base timestamp
base_time = int(datetime(2024, 1, 1, 9, 30).timestamp())

# Generate Nash equilibrium exit times
exits = lucas_encoder.encode_nash_exit_times(base_time, num_exits=5)

print("Nash Equilibrium Exit Times:")
print("=" * 60)
for exit in exits:
    print(f"Exit {exit['exit_index']}: "
          f"L({exit['lucas_index']}) = {exit['lucas_days']} days, "
          f"Days from entry: {exit['days_from_entry']}")

print(f"\n✅ All timestamps are integers")
```
